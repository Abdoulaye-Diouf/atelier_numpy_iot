## Partie 1 : Listes Python vs tableaux NumPy

On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0, 25.8

1) Créeation d'une liste Python stockant ces températures

```python
temp_list = [24.5, 25.1, 26.3, 27.0, 25.8]
print("Températures enregistrées :", temp_list)
```

2) Création d'un tableau NumPy équivalent stockant ces températures

```python
# importation de la bibliothèque NumPy
import numpy as np

#tableau de températures sous forme de tableau NumPy
temp_array = np.array(temp_list)
print("Tableau NumPy des températures :", temp_array)
```

3) Une multiplication par 2 sur ces deux collections

```python
# multiplication par 2 sur ces deux collections de données
 
# multiplication par 2 sur la liste
result_list = temp_list * 2
print("Résultat de la multiplication par 2 sur la liste :", result_list)

# multiplication par 2 sur le tableau NumPy
result_array = temp_array * 2   
print("Résultat de la multiplication par 2 sur le tableau NumPy :", result_array)
```

4) La difference qu'on a observe 

- Sur la liste Python, l'opérateur * réalise une répétition : la liste est dupliquée et concaténée deux fois (elle passe de 5 à 10 éléments), ce qui n'a aucun sens mathématique pour des mesures physiques.
- Sur le tableau NumPy, l'opérateur * réalise une multiplication élément par élément (opération vectorisée) : chaque température est bien multipliée par 2, et la taille du tableau reste inchangée (5 éléments).
- Cette différence illustre l'intérêt majeur de NumPy pour le calcul scientifique : les opérations sont vectorisées, plus rapides et sémantiquement correctes pour des données numériques.

## Partie 8 : Calcul scientifique

Considérons le tableau représentant les 24 températures du 1er bâtiment.

1) Pour les températures calculons le minimum, le maximum, la somme, la moyenne, la médiane, la 
variance et l’écart-type 

```python
 # Calcul du minimum en 24h
minimum_temp = np.min(matrice_capteurs)
print("Minimum des températures (24h) :", f"{minimum_temp:.2f}")

# Calcul du maximum en 24h
maximum_temp = np.max(matrice_capteurs)
print("Maximum des températures (24h) :", f"{maximum_temp:.2f}")

# Calcul de la somme en 24h
somme_temp = np.sum(matrice_capteurs)
print("Somme des températures (24h) :", f"{somme_temp:.2f}")

# Calcul de la moyenne en 24h
moyenne_temp = np.mean(matrice_capteurs)
print("Moyenne des températures (24h) :", f"{moyenne_temp:.2f}")

# Calcul de la médiane en 24h
mediane_temp = np.median(matrice_capteurs)
print("Médiane des températures (24h) :", f"{mediane_temp:.2f}")

# Calcul de la variance en 24h
variance_temp = np.var(matrice_capteurs)
print("Variance des températures (24h) :", f"{variance_temp:.2f}")

# Calcul de l’écart-type en 24h
ecart_type_temp = np.std(matrice_capteurs)
print("Écart-type des températures (24h) :", f"{ecart_type_temp:.2f}")
```

2) Standardisons le tableau des températures puis vérifier les valeurs de la moyenne et de l’écart
type du tableau standardisé 

```python
# 2) Standardise le tableau des températures puis vérifier les valeurs de la moyenne et de l’écart-type du tableau standardisé 
temp_array_standardized = (matrice_capteurs - moyenne_temp) / ecart_type_temp
print("Tableau standardisé des températures :\n", temp_array_standardized)

# verification de la moyenne et de l’écart-type du tableau standardisé
 
print("Vérification -> moyenne :", abs(round(temp_array_standardized.mean(), 10)))
print("Vérification -> écart-type :", round(temp_array_standardized.std(), 10))
```

3) Normalisation (Min-Max) le tableau des températures puis vérifions les valeurs du tableau normalisé

```python
# (Min-Max) le tableau des températures puis vérifier les valeurs du tableau normalisé
temp_array_normalized = (matrice_capteurs - minimum_temp) / (maximum_temp - minimum_temp)
print("Tableau normalisé des températures :\n", temp_array_normalized)

# Vérification des valeurs du tableau normalisé
print("Vérification -> minimum :", round(temp_array_normalized.min(), 10))
print("Vérification -> maximum :", round(temp_array_normalized.max(), 10))
```

4) Convertir le tableau initial en Fahrenheit 

```python
# le tableau initial en Fahrenheit 
temp_array_fahrenheit = (matrice_capteurs * 9/5) + 32
print("Tableau des températures en Fahrenheit :\n", temp_array_fahrenheit)
```
